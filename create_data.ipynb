```python
#load the targets list
import pickle
with open("targets.pkl", "rb") as f:
    targets = pickle.load(f)
```

```python
import data_calibration
import pandas as pd

results = None
for t in targets:
    #print(f"Processing {t.oid}...")
    try:
        df = data_calibration.process_target(t)
        if results is None:
            results = df
        else:
            results = pd.concat([results, df], ignore_index=True)
    except Exception as e:
        pass#print(f"Error processing {t.oid}: {e}")
```

```python
results
```

```python
#plot in log scale the E_exp and M_Ni columns
import matplotlib.pyplot as plt
plt.figure(figsize=(10, 6))
plt.scatter(results["E_exp"], results["M_Ni"], alpha=0.5)
plt.xscale("log")
plt.yscale("log")
plt.xlabel("E_exp (erg)")
plt.ylabel("M_Ni (Msun)")
plt.ylim(1e-2, 3e1)
plt.xlim(1e49, 1e52)
plt.title("E_exp vs M_Ni")
plt.grid(True, which="both", ls="--", lw=0.5)
plt.show()
```

```python
plt.plot(results["E_exp"], results["tp"], "o", alpha=0.5)
plt.xscale("log")
plt.yscale("log")
plt.xlabel("E_exp (erg)")
plt.ylabel("tp (days)")
plt.ylim(50, 2e2)
plt.xlim(1e49, 1e52)
plt.title("E_exp vs tp")
plt.grid(True, which="both", ls="--", lw=0.5)
plt.show()
```

```python
results.to_csv("calibrated_results.csv", index=False)
```
